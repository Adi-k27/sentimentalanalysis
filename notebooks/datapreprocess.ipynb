```python
import pandas as pd

def filter_csv(input_file, output_file, chunksize=100000):
    # Open output file in write mode and write the header once
    first_chunk = True
    
    for chunk in pd.read_csv(input_file, chunksize=chunksize):
        if 'LEVEL1ID' not in chunk.columns:
            print("Error: 'LEVEL1ID' column not found.")
            return
        
        # Filter for Health Canada (LEVEL1ID == 6)
        filtered_chunk = chunk[chunk['LEVEL1ID'] == 6]
        
        # Append to output file
        filtered_chunk.to_csv(output_file, mode='a', header=first_chunk, index=False)
        first_chunk = False  # Ensure only the first chunk writes the header
    
    print(f"Filtered data saved to {output_file}")

# Example usage
input_csv = "../data/dataset/Main Subset.csv"  
output_csv = "../data/clean/mainsubset.csv"
filter_csv(input_csv, output_csv)
```

```python
import pandas as pd
import glob

# Define the directory path where your CSV files are located
directory_path = '../data/processed/'  # Replace with the path to your folder

# Use glob to find all CSV files starting with "subset"
csv_files = glob.glob(f"{directory_path}subset*.csv")

# List to store the results
results = []

# Loop through each CSV file
for file in csv_files:
    print(f"Processing file: {file}")
    data = pd.read_csv(file)
    
    # Check if 'descrip_E' and 'BYCOND' columns exist
    if 'descrip_E' in data.columns and 'BYCOND' in data.columns:
        # Extract unique combinations of descrip_E and BYCOND
        unique_values = data[['descrip_E', 'BYCOND']].drop_duplicates()
        unique_values['source_file'] = file  # Add the file name to the results
        results.append(unique_values)
    else:
        print(f"The required columns 'descrip_E' or 'BYCOND' are missing in {file}")

# Combine results from all files into a single DataFrame
final_result = pd.concat(results, ignore_index=True)

# Save the results to a new CSV file
output_file = 'unique_values.csv'  # Specify your output file name
final_result.to_csv(output_file, index=False)

print(f"Unique values saved to {output_file}")
```

```python
import pandas as pd
import glob
import os

# Define paths
input_directory = '../data/processed/'  # Path where subset CSVs are stored
output_directory = '../data/updated_subsets/'  # Path to save updated files
unique_file = 'unique_values.csv'  # Path to the unique values file

# Create output directory if it doesn't exist
os.makedirs(output_directory, exist_ok=True)

# Load the unique values file
unique_data = pd.read_csv(unique_file)

# Ensure BYCOND column has no leading/trailing spaces and is converted to strings
unique_data["BYCOND"] = unique_data["BYCOND"].astype(str).str.strip()

# Find all subset CSV files in the input directory
csv_files = glob.glob(os.path.join(input_directory, "subset*.csv"))

# Process each subset file
for file in csv_files:
    print(f"Processing file: {file}")
    
    # Load the subset data
    subset_data = pd.read_csv(file)
    
    # Ensure BYCOND column exists in subset data
    if "BYCOND" in subset_data.columns:
        # Apply condition: Trim string if BYCOND has a value and is a string
        # FACTORS column is updated to "General" if BYCOND is empty or invalid
        subset_data["BYCOND"] = subset_data["BYCOND"].astype(str).str.strip()
        
        # Merge with unique_data on BYCOND
        merged_data = subset_data.merge(unique_data[['BYCOND', 'FACTORS']], on='BYCOND', how='left')
        
        # Update FACTORS column to "General" for rows where BYCOND is empty or null
        merged_data["FACTORS"] = merged_data.apply(
            lambda row: "General" if not row["BYCOND"] or row["BYCOND"].strip() == "" else row["FACTORS"],
            axis=1
        )

        # Define output file path
        output_file = os.path.join(output_directory, os.path.basename(file))

        # Save the updated subset file to the new location
        merged_data.to_csv(output_file, index=False)
        print(f"Updated file saved: {output_file}")
    else:
        print(f"Skipping file {file} as it does not contain 'BYCOND' column.")

print("All subset files have been updated and saved in the new location.")
```

```python
import pandas as pd
import glob
import os

# Define paths
input_directory = '../data/updated_subsets/'  # Path where subset CSVs are stored
output_directory = '../data/powerbi/'  # Path to save updated files
question_type_file = 'questions_type.csv'  # Path to the file with QUESTION to QUESTION_TYPE mapping

# Create output directory if it doesn't exist
os.makedirs(output_directory, exist_ok=True)

# Load question type mapping file (QUESTION -> QUESTION_TYPE mapping)
question_type_data = pd.read_csv(question_type_file)

# Convert QUESTION column to string and strip spaces for proper merging
question_type_data["QUESTION"] = question_type_data["QUESTION"].astype(str).str.strip()

# Find all subset CSV files in the input directory
csv_files = glob.glob(os.path.join(input_directory, "subset*.csv"))

# Process each subset file
for file in csv_files:
    print(f"Processing file: {file}")
    
    # Load the subset data
    subset_data = pd.read_csv(file)
    
    # Ensure QUESTION column exists before merging
    if "QUESTION" in subset_data.columns:
        subset_data["QUESTION"] = subset_data["QUESTION"].astype(str).str.strip()

        # Merge with question_type_data to add QUES_TYPE
        subset_data = subset_data.merge(question_type_data[['QUESTION', 'QUESTION_TYPE']], on='QUESTION', how='left')

        # Rename QUESTION_TYPE to QUES_TYPE
        subset_data.rename(columns={"QUESTION_TYPE": "QUES_TYPE"}, inplace=True)

        # Fill missing QUES_TYPE values with "Unknown" (optional)
        subset_data["QUES_TYPE"].fillna("Unknown", inplace=True)
    
    # Define output file path
    output_file = os.path.join(output_directory, os.path.basename(file))

    # Save the updated subset file to the new location
    subset_data.to_csv(output_file, index=False)
    print(f"Updated file saved: {output_file}")

print("All subset files have been updated and saved in the new location.")
```
